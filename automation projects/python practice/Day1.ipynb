## 3. Find Duplicate Elements in List
Input: [1,2,3,2,4,5,1]  \
Output: [1,2]

```python
# solution 1

l = [1,2,3,2,4,5,1]
dupe = []
for i in range(len(l)):
    y = l.pop()
    if y in l:
        dupe.append(y)
dupe
```

```python
# solution 2

l = [1,2,3,2,4,5,1]
list(set([i for i in l if l.count(i)>1]))
```

## 7. Check Palindrome
Input: "madam"   \
Output: True  \
Input: "data"  \
Output: False  

```python
def palindrome(word):
    if word == word[::-1]:
        return True
    else:
        return False
word1 = 'madam'
word2 = 'data'
print(word1, palindrome(word1))
print(word2, palindrome(word2))
```

## 9. Word Count in Sentence
Input:
"data engineering is fun"  \

{

'is':1,
'fun':1
}

```python
# solution 1

word = "data engineering is fun"
array = word.split(' ')
counts = {}
for i in array:
    count = 0
    for j in array:
        if i == j:
            count+= 1
    counts[i] = count
counts
```

## 11. Merge Two Dictionaries
dict1 = {'a':1,'b':2}
dict2 = {'c':3,'d':4}

Output:
{'a':1,'b':2,'c':3,'d':4}

```python
# solution 1

dict1 = {'a':1,'b':2} 
dict2 = {'c':3,'d':4}
temp = {}
def extractor(dic):
    for i,j in dic.items():
        temp[i] = j
extractor(dict1)
extractor(dict2)
temp
```

```python
# solution 2

temp = {**dict1, **dict2}
temp
```

```python
# solution 3

temp = {}
temp.update(dict1)
temp.update(dict2)
temp
```

## 18. Check if Two Strings are Anagrams
Input:
"listen"
"silent"

Output:
True

```python
# solution 1

word1 = 'listen'
word2 = 'silent'

def anagrams(word1, word2):
    if len(word1) == len(word2):
        for i in word1:
            if i in word2:
                continue
            else:
                return False
        return True
anagrams(word1,word2)
        
```

```python
# solution 2

def anagrams(word1, word2):
    return sorted(word1) == sorted(word2)
anagrams(word1,word2)
```

## 19. Generate Fibonacci Series
Input: 7

Output:
0 1 1 2 3 5 8

```python
array = []
for i in range(7):
    if not array:
         array.append(0)
    elif len(array) == 1:
        array.append(1)
    else:
        pattern = array[-1]+array[-2]
        array.append(pattern)
       
' '.join(map(str,array))
```

## 20. Find All Pairs with Target Sum

Input:
nums = [2,7,11,15]
target = 9

Output:
[(2,7)]

```python
nums = [2,7,11,15] 
target = 9
[i for i in nums if target-i in nums] 
```

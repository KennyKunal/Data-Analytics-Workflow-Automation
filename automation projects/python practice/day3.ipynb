## 1. Read a Text File

Create a file data.txt with some lines.
Write a program to read and print the file content.

## 6. Write to a File

Write a program that creates a file and writes 5 lines.

Example output file:

Python  \
Data Engineering  \
ETL  \
Spark  \
Airflow  

Concepts:

"w" mode

```python
with open('test.txt', 'r+') as f:
    l = ['Python\n','Data Engineering\n','ETL\n','Spark\n','Airflow']
    for i in l:
        f.write(i)
    with open('test.txt','r') as fi:
        print(fi.read())
```

## 9. Remove Empty Lines from File

Example file:

engineering

pipeline

Output file:

data  \
engineering  \
pipeline

```python
with open('data.txt','r') as f:
    l = [i for i in f.readlines() if i!='\n']
    
with open('output.txt', 'w') as f:
     for j in l:
        f.write(j)
```

## 13. Read CSV File

Example CSV:

name,age,city
Kunal,25,Mumbai
Rahul,28,Pune

Output:

Kunal 25 Mumbai
Rahul 28 Pune

Use:

csv.reader()

```python
import csv
with open('data.csv') as f:
    file = csv.reader(f)
    next(file)
    for i in file:
        print(f'{i[0], i[1], i[2]}')
```

## 14. Convert CSV to Dictionary

Example output:

]

Use:

csv.DictReader()

```python
with open('data.csv') as f:
    file = csv.DictReader(f)
    print(list(file))
```

## 15. Write Data to CSV

Create a CSV file from a list of dictionaries.

Example input:

[
 {'name':'Aman','age':24},
 {'name':'Neha','age':26}
]

```python
with open('data.csv', 'w') as f:
    data = [ {'name':'Aman','age':24}, {'name':'Neha','age':26} ]
    headers = ['name', 'age']
    file = csv.DictWriter(f, fieldnames = headers)
    file.writeheader()
    for i in data:
        file.writerow(i)
    
```

## 16. Filter CSV Data

Example CSV:

name,age
Aman,24
Neha,30
Rahul,28

Output only age > 25

Neha
Rahul

```python
with open('data.csv','r') as f:
    r = csv.reader(f)
    next(r)
    for i,j in r:
        if int(j)>25:
            print(i)
```

## 18. Convert Dictionary to JSON

Input:

{
"name":"Aman",
"age":26
}

Write it to a JSON file.

Use:

json.dump()

```python
data = '{ "name":"Aman", "age":26 }'

with open('data.json', 'w') as f:
    j = json.loads(data)
    json.dump(j, f)
```

## 19. Convert CSV → JSON

Example CSV:

name,age
Aman,24
Neha,26

Output JSON:

[
 {"name":"Aman","age":24},
 {"name":"Neha","age":26}
]

This is very common in ETL pipelines.

```python
with open('data.csv') as f:
    d = csv.DictReader(f)
    k = list(d)
with open('output.json','w') as f1:
    json.dump(k,f1)
    
```

## 20. JSON Data Filtering

Example JSON:

[
 {"name":"Aman","age":24},
 {"name":"Neha","age":30},
 {"name":"Rahul","age":28}
]

Print only:

age > 25

```python
data = [ {"name":"Aman","age":24}, {"name":"Neha","age":30}, {"name":"Rahul","age":28} ]
for i in data:
    if i['age'] > 25:
        print(i)
```

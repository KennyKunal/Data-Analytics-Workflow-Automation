## 12. Retry Mechanism

Create a function that tries to read a file 3 times before failing.

Example:

Attempt 1 failed
Attempt 2 failed
Attempt 3 failed
File could not be loaded

mThis is common in API or database calls.

```python
def file_reader(filename):
    for i in range(1,4):
        try:
            with open(filename) as f:
                data = f.read()
                return data
        except:
            print(f'attempt {i} failed')
    print('File could not be loaded')

file_reader('dataa.txt')
        
```

## 14. Logging in ETL Pipeline

Create a class:

ETLPipeline

Methods:

extract()
transform()
load()

Log each step.

Example logs:

INFO: Extract started
INFO: Transform completed
ERROR: Load failed

```python
import logging

# Configure logging
logging.basicConfig(level=logging.INFO, format='%(levelname)s: %(message)s')

class ETLPipeline:

    def extract(self):
        logging.info("Extract started")

    def transform(self):
        logging.info("Transform completed")

    def load(self):
        try:
            raise Exception("Load error")
        except Exception:
            logging.error("Load failed")


# Run pipeline
etl = ETLPipeline()
etl.extract()
etl.transform()
etl.load()
```

## 15. Realistic Data Pipeline Error Handling

Simulate pipeline:

Read CSV
Transform Data
Save Output

Handle errors such as:

file missing

invalid data

transformation failure

Log everything.

```python
import logging
import csv

# Configure logging
logging.basicConfig(level=logging.INFO, format='%(levelname)s: %(message)s')

class DataPipeline:

    def extract(self, filename):
        logging.info("Reading CSV file")
        try:
            with open(filename, "r") as f:
                reader = csv.DictReader(f)
                data = list(reader)
            logging.info("File read successfully")
            return data
        except FileNotFoundError:
            logging.error("File missing")
            return None

    def transform(self, data):
        logging.info("Transforming data")
        try:
            for row in data:
                row["age"] = int(row["age"])
                if row["age"] < 0:
                    raise ValueError("Invalid data")
            logging.info("Transformation completed")
            return data
        except Exception:
            logging.error("Transformation failure")
            return None

    def load(self, data):
        logging.info("Saving output")
        try:
            with open("output.csv", "w", newline="") as f:
                writer = csv.DictWriter(f, fieldnames=data[0].keys())
                writer.writeheader()
                writer.writerows(data)
            logging.info("Data saved successfully")
        except Exception:
            logging.error("Load failed")


# Run pipeline
pipeline = DataPipeline()

data = pipeline.extract("input.csv")

if data:
    data = pipeline.transform(data)

if data:
    pipeline.load(data)
```

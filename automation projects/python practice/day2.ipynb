## 3. Function to Find Maximum of Three Numbers
Input: (10, 4, 7)   \
Output: 10

```python
# solution 1

max((10, 4, 7))
```

```python
# solution 2

def maximum(*num):
    m = []
    for i in num:
        m.append(i)
    m.sort()
    return m[-1]

    
maximum(1,2,3)        
```

## 4. Function to Sum All Numbers in List
Input: [1,2,3,4]  \
Output: 10

```python
# solution 1

def sumall(l):
    t = 0
    for i in l:
        t+=i
    return t


sumall([1,2,3,4])
```

```python
# solution 2

array = [1,2,3,4]
from functools import reduce
sumall = reduce(lambda x,y:x+y,array)
sumall
```

## 6. Use *args

Write a function that returns the sum of any number of arguments.

Example:

Input: sum_numbers(1,2,3,4,5)

Output: 15

```python
# solution 1

def sum_numbers(*args):
    temp = 0
    for i in args:
        temp+=i
    return temp

sum_numbers(1,2,3,4,5)
```

```python
# solution 2

sum_num = lambda *args : reduce(lambda x,y:x+y, args)
sum_num(1,2,3,4,5)
```

```python
# solution 3

s = lambda *args: sum(args) 
s(1,2,3,4,5)
```

## 7. Use **kwargs

Create a function that prints key-value pairs.

Example:

person(name="Kunal", age=25, city="Mumbai")

Output:

name : Kunal
age : 25
city : Mumbai

```python
# solution 1

def person(**kwargs):
    return kwargs

person(name='kunal', city='mumbai')
```

```python
# solution 2

person = lambda **kwargs:kwargs
person(name='kunal', city='mumbai')
```

## 14. Recursive Factorial Function
Input: 5
Output: 120

```python
def factorial(n):
    if n == 0:
        return 1
    return n*factorial(n-1)

factorial(5)
```

## 15. Recursive Fibonacci
Input: 6

Output:
0 1 1 2 3 5

```python
def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n-2) + fibonacci(n-1)

n = int(input("Enter n: "))

for i in range(n):
    print(fibonacci(i), end=" ")
```

## 16. Higher Order Function

Create a function that accepts another function as argument.

Example:

apply_function(func, value)

Example call:

apply_function(square, 5)

Output:

25

```python
def square(n):
    return n*n

def cube(n):
    return n**3

def apply_func(func,n):
   return func(n)
print(apply_func(cube,5))
```

## 17. Data Transformation Function

Write a function that converts a list of names to uppercase.

Input:
["kunal","rahul","aman"]

Output:
["KUNAL","RAHUL","AMAN"]

Try using map + lambda.

```python
array = ["kunal","rahul","aman"]

list(map(lambda x:x.upper(),array))
```

```python
def high_case(array):
    return [i.upper() for i in array]

high_case(array)
```

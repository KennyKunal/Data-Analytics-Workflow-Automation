## 1. Create an Iterator

Create a custom iterator that returns numbers from 1 to N.

Example:

Input: 5
Output:
1 2 3 4 5

Concepts:

__iter__()

__next__()

```python
class Numiter:
    def __init__(self,n):
        self.n = n
        self.current = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <=  self.n:
            n = self.current
            self.current += 1
            return n
        else:
            raise StopIteration

num = Numiter(5)

for i in num:
    print(i)
```

## 2. Even Number Iterator

Create an iterator that generates even numbers up to N.

Example:

Input: 10
Output:
2 4 6 8 10

```python
class EvenIter:
    def __init__(self,n):
        self.n = n
        self.current = 2
        
    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= self.n:
            n = self.current
            self.current += 2
            return n
        else:
            raise StopIteration
        

even = EvenIter(10)

for i in even:
    print(i)
        
        
```

## 3. Simple Generator Function

Write a generator function that yields numbers 1 to N.

Example:

Input: 5
Output:
1 2 3 4 5

Concept:

yield

```python
def simple(n):
    for i in range(1,n+1):
        yield i

for j in simple(5):
    print(j)
```

## 4. Fibonacci Generator

Create a generator that yields Fibonacci numbers.

Example:

Input: 6
Output:
0 1 1 2 3 5

```python
def fibo(n):
    a,b = 0,1
    for _ in range(n):
        yield a
        a,b = b,a+b

for i in fibo(6):
    print(i)
        
            
            
```

## 5. Generator for Large File Processing

Create a generator that reads a file line by line.

Example use case:

Process large log files

Hint:

yield line

```python
def read_file(filename):
    with open(filename) as f:
        for i in f:
            yield i

for p in read_file('data.txt'):
    print(p)
```

## 7. Basic Decorator

Create a decorator that prints:

Function started
Function finished

around a function call.

Example:

@log_execution
def process_data():

```python
def log_execution(func):
    def wraper():
        print('Function started')
        a = func()
        print('function fininshed')
        return a
    return wraper

@log_execution
def process_data():
    print('processing data')

process_data()
```

## 8. Execution Time Decorator

Create a decorator that measures execution time of a function.

Example output:

Function took 0.25 seconds

Use:

time module

```python
import time 

def timecal(func):
    def wraper():
        start = time.perf_counter()
        a = func()
        stop = time.perf_counter()
        return f'Function took {stop-start} seconds'
    return wraper

@timecal
def test():
    for i in range(10000000):
        pass
    return i

test()
```

## 9. Retry Decorator

Create a decorator that retries a function 3 times if it fails.

Example output:

Attempt 1 failed
Retrying...
Attempt 2 failed
Retrying...
Attempt 3 failed

```python
def retry(func):
    def wraper():
        attempt = 0
        for _ in range(3):
            try:
                a = func()
                return a
            except:
                attempt += 1
                print(f'Attempt {attempt} failed')
                if attempt < 3:
                    print('Retrying...')
    return wraper

@retry
def test1():
    a = 'abc'
    raise Exception('error')

test1()
```

## 10. Multiprocessing Example

Create a program that calculates square of numbers using multiprocessing.

Example input:

[1,2,3,4,5]

Output:

[1,4,9,16,25]

Concept:

multiprocessing.Pool

```python
import concurrent.futures

def square(n):
    return n*n

if __name__ == '__main__':
    with concurrent.futures.ProcessPoolExecutor() as executor:
        result = executor.map(square,[2,3,4])
        for i in result:
            print(i)
```

## 12. Compare Performance

Compare execution time of:

Normal loop

List comprehension

Generator

Example:

sum of 1 to 1,000,000

```python

@timecal
def normal():
    t = 0
    for i in range(1,1000000):
        t += i
    return t



@timecal
def listcomp():
    return sum(list(range(1,1000000)))

@timecal
def gencomp():
    return sum(i for i in range(1,1000000))

print(normal())
print(listcomp())
gencomp()
```

## 14. Data Pipeline Generator

Create a generator-based pipeline:

read_data()
clean_data()
transform_data()

Example flow:

data → generator → cleaned → transformed

```python

def read_data():
    data = ["  apple ", "banana  ", "  cherry", None, "date  "]
    for item in data:
        yield item

def clean_data(data_gen):
    for item in data_gen:
        if item is None:
            continue
        yield item.strip()

def transform_data(cleaned_gen):
    for item in cleaned_gen:
        yield item.upper()

pipeline = transform_data(clean_data(read_data()))

for item in pipeline:
    print(item)
```

## 15. Decorator for Logging

Create a decorator that logs:

Function name
Arguments
Execution time

Example:

Calling process_data with args (file.csv)
Execution time: 0.12 sec

```python
import logging
import csv
import time
logging.basicConfig(level=logging.INFO)
def timecal(func):
    def wraper(data):
        start = time.perf_counter()
        a = func(data)
        stop = time.perf_counter()
        return logging.info(f'Function took {stop-start} seconds')
    return wraper

@timecal
def csvreader(filename):
    with open(filename) as f:
        file = csv.reader(f)
        time.sleep(1)

csvreader('data.csv')
```

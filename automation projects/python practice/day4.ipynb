## 5. Create a Bank Account Class

Attributes:

account_holder

balance

Methods:

deposit()

withdraw()

check_balance()

Example:

deposit(500)
withdraw(200)

```python
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance
        
    def deposit(self,depo):
        self.balance = self.balance + depo

    def withdraw(self, withdraw):
        self.balance = self.balance + withdraw

    def check_balance(self):
        return self.balance

user1 = BankAccount('user1', 1000)
user1.deposit(500)
print(user1.check_balance())
user1.withdraw(200)
print(user1.check_balance())
```

```python
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.__balance = balance
        
    def deposit(self,depo):
        self.__balance = self.__balance + depo

    def withdraw(self, withdraw):
        self.__balance = self.__balance + withdraw

    def check_balance(self):
        return self.balance

    def get_balance(self):
        return self.__balance
        
```

```python
user1 = BankAccount('user1', 1000)
print(user1.deposit(500))
print(user1.get_balance())
```

## 8. Inheritance Example

Create class:

Animal

Method:

speak()

Create child classes:

Dog
Cat

Example output:

Dog → Bark
Cat → Meow

```python
class Animal:
    def speak(self):
        print('sound')

class Dog(Animal):
    def speak(self):
        print('Bark')

class Cat(Animal):
    def speak(self):
        print('meow')

dog = Dog()
dog.speak()

cat = Cat()
cat.speak()
```

## 12. Create a CSVProcessor Class

Attributes:

file_name

Methods:

read_file()
count_rows()
count_columns()

```python
import csv

class CSVProcessor:
    def __init__(self,filename):
        self.filename = filename
        self.data = []

    def read_file(self):
        with open(self.filename) as f:
            reader = csv.reader(f)
            self.data = list(reader)
    
    def count_rows(self):
        return len(self.data)
        
    def count_columns(self):
        return len(self.data[0])

data = CSVProcessor('data.csv')
data.read_file()
print(data.count_rows())
print(data.count_columns())
```

## 13. Inheritance in Data Processing

Create:

DataSource (Parent Class)

Methods:

connect()
fetch_data()

Child classes:

MySQLSource
APISource
CSVSource

Override fetch_data().

```python
class DataSource():
    def connect(self):
        return 'data connecting'
    def fetch_data(self):
        return 'fetching data'

class MySQLSource(DataSource):
    def fetch_data(self):
        return 'fetching sql data'

class APISource(DataSource):
    def fetch_data(self):
        return 'fetching API'

class CSVSource(DataSource):
    def fetch_data(self):
        return 'fetching CSV'

mysql = MySQLSource()
api = APISource()
csv = CSVSource()

mysql.fetch_data()
api.fetch_data()
csv.fetch_data()
```

## 15. ETL Pipeline Simulation (Mini Project)

Create a class:

ETLProcess

Methods:

extract()  → read CSV
transform() → clean data
load() → print "data loaded"

Example output:

Extracting data...
Transforming data...
Loading to database...

This is very close to real Data Engineering code.

```python
import csv

class ETLProcess:
    def __init__(self,filename):
        self.filename = filename
        self.data = []

    def extract(self):
        with open(self.filename) as f:
            reader = csv.reader(f)
            self.data = list(reader)

    def transform(self):
        self.data = [row for row in self.data if row]

    def load(self):
        print('loading to database...')

etl = ETLProcess('data.csv')

etl.extract()
etl.transform()
etl.load()
```
